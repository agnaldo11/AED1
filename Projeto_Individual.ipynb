```python
import random

# Tupla - Tipos de caracteres permitidos (fixos)
tipos_disponiveis = ("minusculas", "maiusculas", "numeros", "simbolos")

# Dicionário - Cada chave representa um tipo de caractere
conjuntos = {
    "minusculas": "abcdefghijklmnopqrstuvwxyz",
    "maiusculas": "ABCDEFGHIJKLMNOPQRSTUVWXYZ",
    "numeros": "0123456789",
    "simbolos": "!@#$%&*?"
}

# Função principal paraa gera a senha
def gerar_senha(tamanho, usar_maiusculas, usar_numeros, usar_simbolos):

    # Lista onde serão armazenados os caracteres permitidos
    caracteres = []

    # 1) Minúsculas sempre entram
    for ch in conjuntos["minusculas"]:
        caracteres.append(ch)

    # 2) Condicionais para adicionar outros tipos
    if usar_maiusculas:
        for ch in conjuntos["maiusculas"]:
            caracteres.append(ch)

    if usar_numeros:
        for ch in conjuntos["numeros"]:
            caracteres.append(ch)

    if usar_simbolos:
        for ch in conjuntos["simbolos"]:
            caracteres.append(ch)

    # Proteção contra lista vazia
    if not caracteres:
        return "Erro: nenhum tipo de caractere selecionado."

    # 3) Gerar senha usando loop
    senha = " "
    for _ in range(tamanho):
        senha += random.choice(caracteres)

    return senha

tam = int(input("Tamanho da senha: "))
mai = input("Incluir MAIÚSCULAS? (s/n): ").lower() == "s"
num = input("Incluir Numeros? (s/n): ").lower() == "s"
sim = input("Incluir Símbolos? (s/n): ").lower() == "s"

print("\nSenha gerada:", gerar_senha(tam, mai, num, sim))
```

```python
n = int(input("Digite o valor de n :"))
fat = 1
if n < 0:
    print("fatirial não existe!")
elif n == 0:
    print("fatirial é igual a 1!")
else: 
    for i in range(1, n+1):
        fat = fat * i
print("o fatorial de " , n, "é igual a ", fat)        
```

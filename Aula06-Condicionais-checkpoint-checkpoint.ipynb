Os operadores lógicos funcionam assim:

- Operador **and** - Retorna True se ambas as declarações forem verdadeiras.
- Operador **or** - Retorna True se uma das declarações for verdadeira.
- Operador **not** - Inverte o resultado, retorna False se o resultado for True.

```python
# Operador and
numero = 4

if (numero > 2) and (numero % 2 == 0):
    print("Isso está sendo impresso porque as duas condições são verdadeiras!")
```

```python
# Operador and
numero = 4

if (numero > 5) and (numero % 2 == 0):
    print("Isso está sendo impresso porque as duas condições são verdadeiras!")
else:
    print("Isso está sendo impresso porque uma das duas condições é falsa!")
```

```python
# Operador or

numero = 4

if (numero > 5) or (numero % 2 == 0):
    print("Isso está sendo impresso porque uma duas condições é verdadeira!")
```

```python
# Operador not

numero = 4

if not(numero > 5) and (numero % 2 == 0):
    print("Isso está sendo impresso porque as duas condições são verdadeiras!")
else:
    print("Isso está sendo impresso porque uma das duas condições é falsa!")
```

```python
# Operador and, or e not

numero = 4

if (not(numero > 5) and (numero % 2 == 0)) or (numero == 4):
    print("Isso está sendo impresso porque as duas primeiras condições são verdadeiras ou a terceira é verdadeira!")
```

```python
# gerador_senha_simples.py

import random  # usado para escolher caracteres aleatórios

def gerar_senha(tamanho, incluir_maiusculas, incluir_numeros, incluir_simbolos):
    """
    Gera uma senha aleatória com base nas opções fornecidas.
    Parâmetros:
      - tamanho: int (quantos caracteres)
      - incluir_maiusculas: bool
      - incluir_numeros: bool
      - incluir_simbolos: bool
    Retorna:
      - senha (str)
    """

    # 1) Conjuntos de caracteres (strings)
    minusculas = "abcdefghijklmnopqrstuvwxyz"
    maiusculas = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    numeros = "0123456789"
    simbolos = "!@#$%&*?"

    # 2) Lista para juntar caracteres permitidos
    caracteres_possiveis = []  # lista vazia (mutável)

    # 3) Sempre adicionamos minúsculas — assim garantimos pelo menos um tipo
    for ch in minusculas:
        caracteres_possiveis.append(ch)

    # 4) Condicionais para adicionar outros tipos (se solicitado)
    if incluir_maiusculas:
        for ch in maiusculas:
            caracteres_possiveis.append(ch)

    if incluir_numeros:
        for ch in numeros:
            caracteres_possiveis.append(ch)

    if incluir_simbolos:
        for ch in simbolos:
            caracteres_possiveis.append(ch)

    # 5) Proteção simples: se por algum motivo a lista ficar vazia, lançamos erro
    if not caracteres_possiveis:
        raise ValueError("Nenhum tipo de caractere selecionado.")

    # 6) Montagem da senha: loop 'for' para escolher 'tamanho' caracteres
    senha = ""  # string (imutável) que iremos concatenar
    for _ in range(tamanho):
        senha += random.choice(caracteres_possiveis)

    # 7) Retorna a senha gerada
    return senha


# Exemplos de uso (para você rodar localmente no vídeo)
# senha1 = gerar_senha(8, False, True, False)
# senha2 = gerar_senha(12, True, True, True)
```

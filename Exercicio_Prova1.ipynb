```python
# Escreva um programa que leia o ano de nascimento e o ano atual, e calcule a idade em anos, meses, dias e semanas.
ano_nascimento = int(input('Digite o ano de nascimento: '))
ano_atual = int(input('Digite o ano atual: '))
idade_ano = ano_atual - ano_nascimento
idade_meses = idade_ano * 12
idade_dias = idade_ano * 365
idade_semana = idade_dias // 7
print(f'Idade: {idade_ano} anos, {idade_meses} meses, {idade_dias} dias')
```

```python
# Escreva um programa que calcule o novo salário após um aumento de 25%.
salario = int(input('Digite seu salario: '))
salario_atual = salario * 1.25
print(f' Salario atual: {salario_atual}')
```

```python
# Escreva um programa que calcule a área de um círculo, dado o raio.
raio = float(input('Digite o raio: '))
area = 3.14 * (raio ** 2)
print(f'Área do circulo: {area}')
```

```python
# Escreva um programa que calcule quantos degraus uma pessoa precisa subir para alcançar uma certa altura.
altura_degraus = float(input('Digite a altura do degrau: '))
altura_desejada = float(input('Digite altura do degrau desejada: '))
num_degraus = (altura_desejada * 100) / altura_degraus
print(f'Altura que você precisa subir: {num_degraus:.0f} degraus')
```

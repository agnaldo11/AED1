```python
valor1 = int(input('Digite o primeiro valor: '))
valor2 = int(input('Digite o segundo valor: '))
valor3 = int(input('Digite o terceiro valor: '))

soma = (valor1 + valor2 + valor3)

menor = valor1

if valor2 < menor:
  menor = valor2

if valor3 < menor:
  menor = valor3

media = (soma - menor) / 2

print(f'Media aritimetica dos dois valor: {media}')
```
